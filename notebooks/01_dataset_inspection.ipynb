# SECFSN+ Dataset Inspection

This research notebook inspects the demo modeling dataset using the repository's existing notebook and pipeline helpers.


## 1. Load Data


```python
from secfsn_plus_models.notebook_utils import (
    load_notebook_demo_artifacts,
)

demo = load_notebook_demo_artifacts(
    model_type="linear_regression",
    train_end_period="2020Q4"
)

X = demo.X
y = demo.y

print(f"Dataset source: {demo.dataset_source}")
```

## 2. Basic Dataset Summary


```python
print(f"X.shape: {X.shape}")
print(f"y.shape: {y.shape}")
print(f"Number of features: {X.shape[1]}")
print("Feature column names:")
for column in X.columns:
    print(f"- {column}")

cik_count = X.index.get_level_values("cik").nunique()
period_count = X.index.get_level_values("period").nunique()
print(f"Unique cik count: {cik_count}")
print(f"Unique period count: {period_count}")
```

## 3. Period Coverage


```python
import matplotlib.pyplot as plt

rows_per_period = X.groupby(level="period", sort=False).size()
print(rows_per_period)
```

```python
fig, ax = plt.subplots(figsize=(10, 4))
rows_per_period.plot(kind="bar", ax=ax)
ax.set_title("Rows per Period")
ax.set_xlabel("Period")
ax.set_ylabel("Row Count")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## 4. Label Distribution


```python
fig, ax = plt.subplots(figsize=(8, 4))
ax.hist(y, bins=20)
ax.set_title("Label Distribution")
ax.set_xlabel(y.name or "label")
ax.set_ylabel("Frequency")
plt.tight_layout()
plt.show()
```

```python
print(f"Mean: {y.mean():.6f}")
print(f"Std: {y.std():.6f}")
print(f"Min: {y.min():.6f}")
print(f"Max: {y.max():.6f}")
```

## 6. Feature Correlations


```python
import pandas as pd

correlation_matrix = X.corr()
correlation_pairs = []
for left_index, left_feature in enumerate(correlation_matrix.columns):
    for right_feature in correlation_matrix.columns[left_index + 1:]:
        correlation_pairs.append(
            {
                "feature_1": left_feature,
                "feature_2": right_feature,
                "correlation": correlation_matrix.loc[left_feature, right_feature],
                "abs_correlation": abs(correlation_matrix.loc[left_feature, right_feature]),
            }
        )

top_correlations = (
    pd.DataFrame(correlation_pairs)
    .sort_values("abs_correlation", ascending=False)
    .head(10)
    .drop(columns="abs_correlation")
    .reset_index(drop=True)
)

top_correlations
```
