# Model Comparison Demo

Compare the baseline linear regression and random forest models on the same SECFSN+ demo split using repository pipeline artifacts.


```python
import pandas as pd

from secfsn_plus_models.notebook_utils import (
    COMPARISON_METRIC_COLUMNS,
    load_notebook_demo_artifacts,
    summarize_model_comparison_metrics,
)

linear_demo = load_notebook_demo_artifacts(
    model_type="linear_regression",
    train_end_period="2020Q4",
)
random_forest_demo = load_notebook_demo_artifacts(
    model_type="random_forest",
    train_end_period="2020Q4",
)

print(f"Dataset source: {linear_demo.dataset_source}")
print(f"Comparison periods: {list(linear_demo.pipeline.ic_by_period.index)}")
```

## Summary Metrics

The table below keeps the requested comparison metrics side by side for both baseline models.


```python
comparison = pd.DataFrame(
    [
        summarize_model_comparison_metrics(
            "linear_regression",
            linear_demo.pipeline.y_test,
            linear_demo.pipeline.predictions,
            ic_by_period=linear_demo.pipeline.ic_by_period,
            rank_ic_by_period=linear_demo.pipeline.rank_ic_by_period,
        ).to_dict(),
        summarize_model_comparison_metrics(
            "random_forest",
            random_forest_demo.pipeline.y_test,
            random_forest_demo.pipeline.predictions,
            ic_by_period=random_forest_demo.pipeline.ic_by_period,
            rank_ic_by_period=random_forest_demo.pipeline.rank_ic_by_period,
        ).to_dict(),
    ]
)
comparison = comparison.set_index("model_name")[COMPARISON_METRIC_COLUMNS]
comparison.round(4)
```

## IC by Period

Plot the period-level information coefficient for both models on the shared evaluation window.


```python
import matplotlib.pyplot as plt

ic_by_period = pd.concat(
    [
        linear_demo.pipeline.ic_by_period.rename("linear_regression"),
        random_forest_demo.pipeline.ic_by_period.rename("random_forest"),
    ],
    axis=1,
)

ax = ic_by_period.plot(marker="o")
ax.set_title("IC by Period")
ax.set_xlabel("Period")
ax.set_ylabel("IC")
ax.axhline(0.0, linestyle="--")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
